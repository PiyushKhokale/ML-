```python
import pandas as pd
```

```python
data = pd.read_csv("enjoy.csv")
```

```python
X = data.iloc[:, :-1].values
Y = data.iloc[:, -1].values
```

```python
import numpy as np


def find_s(X, Y):
    hypothesis = None
    print("Initial Hypothesis :", hypothesis)
    for i in range(len(X)):
        if Y[i] == "Yes":
            if hypothesis is None:
                hypothesis = X[i].copy()
            else:
                for j in range(len(hypothesis)):
                    if hypothesis[j] != X[i][j]:
                        hypothesis[j] = "?"
        print("After Training Example {i+1}")
        print(hypothesis)
    return hypothesis


final_hypothesis = find_s(X, Y)
print("Final Hypothesis :", final_hypothesis)
```

```python
def candidate_elimination(concepts, target):
    specific = concepts[0].copy()
    general = [["?" for i in range(len(specific))]]
    print("Initial S:", specific)
    print("Initial G:", general)
    for i, h in enumerate(concepts):
        if target[i] == "Yes":
            for x in range(len(specific)):
                if h[x] != specific[x]:
                    specific[x] = "?"
                    general[x][x] = "?"
                return specific, general


S, G = candidate_elimination(X, Y)
print("Final Specific Hypothesis:")
print(S)
print("Final General Hypothesis:")
print(G)
for g in G:
    print(g)
```
