```python
import pandas as pd
import xml.etree.ElementTree as ET
import json
```

# 1. Data Collection

## Txt File

```python
with open("data/students.txt","w") as f:
    f.write("ID Name Age\n")
    f.write("1 Saptanshu 22\n")
    f.write("2 Hardik 21\n")
    f.write("3 Saumya 22")
    f.write("4 Piyush 22")
```

## JSON File

```python
students = [
    {"id": 1, "name": "Saptanshu", "age": 22},
    {"id": 2, "name": "Hardik", "age": 21},
    {"id": 3, "name": "Saumya", "age": 22},
    {"id": 4, "name": "Piyush", "age": 22}
]

with open("data/students.json","w") as f:
    json.dump(students, f, indent=4)
```

## XML File

```python
root = ET.Element("students")

for student in students:
    s = ET.SubElement(root, "student")
    ET.SubElement(s, "id").text = str(student["id"])
    ET.SubElement(s, "name").text = student["name"]
    ET.SubElement(s, "age").text = str(student["age"])

tree = ET.ElementTree(root)
tree.write("data/students.xml")
```

# 3. Data Analysis

```python
customers = pd.DataFrame({
    "CustomerID": [101, 102, 103, 104],
    "Name": ["Saptanshu", "Hardik", None, "Saumya"],
    "Age": [22, 21, None, 22],
    "City": ["Delhi", "Mumbai", "Delhi", None]
})
```

```python
orders = pd.DataFrame({
    "OrderID": [1, 2, 3, 4],
    "CustomerID": [101, 102, 101, 104],
    "Amount": [500, 700, None, 450]
})
```

```python
print("\nCustomers Data")
print(customers)

print("\nOrders Data")
print(orders)

print("\nCustomers Info")
customers.info()

print("\nOrders Info")
orders.info()
```

```python
customers["Name"] = customers["Name"].fillna("Unknown")
```

```python
customers["Age"] = customers["Age"].fillna(customers["Age"].mean())
```

```python
customers["City"] = customers["City"].fillna("Unknown")
```

```python
orders["Amount"] = orders["Amount"].fillna(orders["Amount"].mean())
```

```python
customers = customers.drop_duplicates()
orders = orders.drop_duplicates()
```

```python
merged = pd.merge(customers, orders, on="CustomerID")
```

```python
print("\nMerged Data")
print(merged)

print("\nMerged Info")
merged.info()

print("\nMerged Describe")
merged.describe()
```
