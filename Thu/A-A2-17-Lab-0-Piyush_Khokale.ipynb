```python
print("\n--- while loop ---")
print("The while loop executes a block of code as long as a condition is true.")

count = 0
print(f"Counting upto 3 using while loop:")
while count < 3:
    print(f"Count: {count}")
    count += 1

    secret_number = 7
    guess = 0
    print(f"\nGuess the secret number (between 1 and 10):")
    if guess == secret_number:
        print(f"Congratulations! You guessed the secret number {secret_number}.")
        break
    elif guess > secret_number + 2:
        print(f"Too many wrong guesses. stopping")
        break
    else:
        print(f"Guess {guess} : Not correct. Try again.")
```

```python
def power(base, exp=2):
    """Function to calculate power of a number with a default exponent."""
    return base ** exp

print("\n Function with default argument")
print(f"5 raised to the power of 2 (default): {power(5)}")
print(f"5 raised to the power of 3: {power(5 , 3)}")

print("\n Recursion in python")
```

```python
def factorial(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n * factorial(n-1)
    
print(f"factorial using recursion")
num = 5 
print(f"The factorial of {num} is: {factorial(num)}")

num = 0 
print(f"The factorial of {num} is: {factorial(num)}")
```

```python
def fibonacci(n):
    if n <= 0:
        return []
    elif n == 1:
        return [0]
    elif n == 2:
        return [0, 1]
    else:
        fib_sequence = fibonacci(n - 1)
        fib_sequence.append(fib_sequence[-1] + fib_sequence[-2])
        return fib_sequence
    
fibonacci(10)
```
